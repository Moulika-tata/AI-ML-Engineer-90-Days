## 2. Simplifying Fractions Using HCF

To simplify a fraction:

1. Find the HCF of the numerator and denominator.
2. Divide both by the HCF.

Example:

\[
\frac{18}{24}
\]

HCF of 18 and 24 = 6.

\[
\frac{18\div6}{24\div6}=\frac{3}{4}
\]

Therefore, the simplest form is \(\frac{3}{4}\).

```python

# Simplify a fraction using the HCF

import math

numerator = 18
denominator = 24

hcf = math.gcd(numerator, denominator)

simplified_numerator = numerator // hcf
simplified_denominator = denominator // hcf

print("Original fraction:", f"{numerator}/{denominator}")
print("HCF:", hcf)
print("Simplified fraction:",
      f"{simplified_numerator}/{simplified_denominator}")
```

## 3. Operations on Fractions

### Addition and Subtraction

If denominators are the same, add or subtract the numerators.

\[
\frac{2}{5}+\frac{1}{5}=\frac{3}{5}
\]

If denominators are different, find their LCM and convert the fractions first.

\[
\frac{1}{2}+\frac{1}{3}
=\frac{3}{6}+\frac{2}{6}
=\frac{5}{6}
\]

### Multiplication

Multiply the numerators together and the denominators together.

\[
\frac{3}{4}\times\frac{2}{5}=\frac{6}{20}=\frac{3}{10}
\]

### Division

Keep the first fraction, change division to multiplication, and flip the second fraction.

\[
\frac{1}{2}\div\frac{1}{4}
=\frac{1}{2}\times\frac{4}{1}=2
\]

```python

# Fraction operations using Python's Fraction class

from fractions import Fraction

a = Fraction(1, 2)
b = Fraction(1, 3)

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)

# Another example
c = Fraction(3, 4)
d = Fraction(2, 5)

print("3/4 * 2/5 =", c * d)
```

## 4. Converting Fractions and Decimals

### Fraction to Decimal

Divide the numerator by the denominator.

\[
\frac{3}{5}=3\div5=0.6
\]

### Decimal to Fraction

Write the decimal over 10, 100, or 1,000, depending on the number of decimal places. Then simplify.

\[
0.75=\frac{75}{100}=\frac{3}{4}
\]

\[
0.625=\frac{625}{1000}=\frac{5}{8}
\]

### Recurring Decimals

A recurring decimal has a digit or group of digits that repeats forever.

\[
0.3333\ldots=\frac{1}{3}
\]

\[
0.6666\ldots=\frac{2}{3}
\]

```python

# Convert fractions to decimals and decimals to fractions

from fractions import Fraction

# Fraction to decimal
fraction = Fraction(3, 5)
print("3/5 as a decimal:", float(fraction))

# Decimal to fraction
decimal_1 = Fraction("0.75")
decimal_2 = Fraction("0.625")
decimal_3 = Fraction("0.4")

print("0.75 as a fraction:", decimal_1)
print("0.625 as a fraction:", decimal_2)
print("0.4 as a fraction:", decimal_3)

# Recurring decimal equivalents
print("1/3 as a decimal:", float(Fraction(1, 3)))
print("2/3 as a decimal:", float(Fraction(2, 3)))
```
