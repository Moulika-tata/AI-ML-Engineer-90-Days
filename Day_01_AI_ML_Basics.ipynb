## Exercise 5 — Prime Number

A prime number has exactly two factors:

1. 1
2. The number itself

Examples:
- 2 → Prime
- 3 → Prime
- 5 → Prime
- 7 → Prime

1 is NOT a prime number.

```python
number = int(input("Enter number: "))

if number < 2:
    print("Not prime")
else:
    is_prime = True

    for i in range(2, number):
        if number % i == 0:
            is_prime = False
            break

    if is_prime:
        print("Prime")
    else:
        print("Not prime")
```

### Prime Number Logic

We initially assume that the number is prime.

If the number can be divided exactly by any number between
2 and number - 1, the remainder will be 0.

Therefore, the number is not prime.

Example:

7 % 2 → 1
7 % 3 → 1
7 % 4 → 3
7 % 5 → 2
7 % 6 → 1

No remainder is 0, so 7 is prime.

For 6:

6 % 2 → 0

Therefore, 6 is not prime.
