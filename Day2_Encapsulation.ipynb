## 1. What is Encapsulation?

Encapsulation is an OOP concept that combines data (attributes) and methods into a class while controlling access to the data.

**Benefits:**
- Protects data from invalid modifications.
- Allows controlled access to attributes.
- Improves code organization and maintainability.

## 2. Types of Attribute Naming in Python

- **Public:** `balance` — can be accessed directly.
- **Protected-style:** `_balance` — conventionally intended for internal use.
- **Private-style:** `__balance` — Python uses name mangling to make accidental access harder.

Note: Python's double underscore does not provide absolute privacy.

## 3. What is `self`?

`self` refers to the current object instance. It allows an object to access its own attributes and methods.

## 4. What is `@property`?

The `@property` decorator allows a method to be accessed like an attribute.

## 5. What is `@setter`?

A setter allows us to validate a value before updating an attribute.

## 6. Real-World Example

An employee's salary should not be negative. Encapsulation allows us to reject invalid salary updates while keeping the existing valid salary unchanged.

```python

# Day 2: OOP Encapsulation

# Example 1: Employee salary validation

class Employee:
    def __init__(self, salary):
        if salary >= 0:
            self.__salary = salary
        else:
            raise ValueError("Salary cannot be negative!")

    @property
    def salary(self):
        return self.__salary

    @salary.setter
    def salary(self, new_salary):
        if new_salary >= 0:
            self.__salary = new_salary
        else:
            print("Invalid salary!")


# Create an employee object
employee1 = Employee(25000)

print("Initial Salary:", employee1.salary)

# Try to set an invalid salary
employee1.salary = -5000
print("Salary after invalid update:", employee1.salary)

# Update with a valid salary
employee1.salary = 30000
print("Salary after valid update:", employee1.salary)


# Example 2: Student marks validation

class Student:
    def __init__(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            raise ValueError("Marks must be between 0 and 100!")

    def update_marks(self, new_marks):
        if 0 <= new_marks <= 100:
            self.__marks = new_marks
            print("Marks updated successfully!")
        else:
            print("Invalid marks!")

    def display_marks(self):
        print("Student Marks:", self.__marks)


student1 = Student(80)

student1.display_marks()

student1.update_marks(95)
student1.display_marks()

student1.update_marks(150)
student1.display_marks()
```
