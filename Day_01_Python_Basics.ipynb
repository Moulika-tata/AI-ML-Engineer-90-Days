## 5. input()

## 6. Type Conversion

Type conversion means changing a value from one data type to another.

Common conversions:

- int() → converts to integer
- float() → converts to floating-point number
- str() → converts to string

```python
age = int(input("Enter your age: "))
cgpa = float(input("Enter your CGPA: "))

print(age)
print(cgpa)
```

## Program 1 — Sum of Two Numbers

Formula:

Sum = number1 + number2

```python
number1 = int(input("Enter first number: "))
number2 = int(input("Enter second number: "))

total = number1 + number2

print("Sum =", total)
```

## Program 2 — Average of Two Numbers

Formula:

Average = (number1 + number2) / 2

```python
number1 = int(input("Enter number1: "))
number2 = int(input("Enter number2: "))

average = (number1 + number2) / 2

print("Average =", average)
```

## Program 3 — Area of a Rectangle

Formula:

Area = length × width

```python
length = int(input("Enter length: "))
width = int(input("Enter width: "))

area = length * width

print("Area =", area)
```

## Program 4 — Celsius to Fahrenheit

Formula:

Fahrenheit = (Celsius × 9/5) + 32

```python
celsius = int(input("Enter Celsius: "))

fahrenheit = (celsius * 9 / 5) + 32

print("Fahrenheit =", fahrenheit)
```

## Program 5 — Simple Interest

Formula:

Simple Interest = (Principal × Rate × Time) / 100

Where:

- Principal = Amount of money
- Rate = Interest rate
- Time = Time period

```python
principal = int(input("Enter principal: "))
rate = int(input("Enter rate: "))
time = int(input("Enter time: "))

simple_interest = (principal * rate * time) / 100

print("Simple Interest =", simple_interest)
```

## Program 6 — Swap Two Values

Swapping means exchanging the values of two variables.

Python allows us to swap two values using:

a, b = b, a

```python
a = int(input("Enter a: "))
b = int(input("Enter b: "))

print("Before swapping:")
print("a =", a)
print("b =", b)

a, b = b, a

print("After swapping:")
print("a =", a)
print("b =", b)
```

## Program 7 — Even or Odd

The modulo operator (%) gives the remainder.

If:

number % 2 == 0

the number is Even.

Otherwise, the number is Odd.

```python
num = int(input("Enter num: "))

if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

## Program 8 — Largest of Two Numbers

We compare two numbers using the greater-than operator (>).

If num1 is greater than num2, num1 is the largest.

Otherwise, num2 is the largest.

```python
num1 = int(input("Enter num1: "))
num2 = int(input("Enter num2: "))

if num1 > num2:
    print("Largest =", num1)
else:
    print("Largest =", num2)
```
