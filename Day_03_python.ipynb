## 1. if Statement

The `if` statement is used to execute a block of code when a condition is True.

### Syntax

if condition:
    statement

Python uses indentation to define the block of code.

```python
age = 16

if age >= 18:
    print("Eligible")
else:
    print("Not Eligible")
```

## 2. if-else Statement

The `if-else` statement provides two possible paths.

- If the condition is True → `if` block executes.
- If the condition is False → `else` block executes.

```python
marks = 75

if marks >= 90:
    print("A")
elif marks >= 60:
    print("B")
else:
    print("C")
```

## 3. if-elif-else Statement

`elif` means "else if".

It is used to check another condition when the previous condition is False.

- `if` → checks the first condition
- `elif` → checks another condition
- `else` → executes when none of the conditions are True

```python
age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Allowed")
    else:
        print("ID Required")
else:
    print("Not Eligible")
```

## 4. Nested if

A nested `if` means placing one `if` statement inside another `if` statement.

It is useful when one condition needs to be checked only after another condition is satisfied.

```python
for i in range(5):
    print(i)
```

## 5. for Loop

A `for` loop is generally used to iterate over a sequence or range.

Example:

for i in range(5):
    print(i)

Output:
0
1
2
3
4

```python
i = 1

while i <= 5:
    print(i)
    i += 1
```

## 6. while Loop

A `while` loop repeats a block of code as long as a condition is True.

The condition is checked before each iteration.

Important:
The loop variable should be updated to avoid an infinite loop.

```python
for i in range(1, 6):
    print(i)
```

## 7. range()

`range()` generates a sequence of numbers.

### Examples

range(5)
→ 0, 1, 2, 3, 4

range(1, 6)
→ 1, 2, 3, 4, 5

The stop value is not included.

```python
for i in range(1, 7):
    if i == 6:
        break
    print(i)
```

## 8. break

The `break` statement is used to stop a loop completely.

When Python reaches `break`, the loop terminates immediately.

```python
for i in range(1, 11):
    if i == 5:
        continue
    print(i)
```

## 9. continue

The `continue` statement skips the current iteration and moves to the next iteration.

### Remember

- `break` → Stop the loop
- `continue` → Skip the current iteration

```python
num = int(input("Enter number: "))

if num % 3 == 0 and num % 5 == 0:
    print("Divisible by both")
else:
    print("Not divisible by both")
```

# 11. Interview Questions

### Q1. What is a conditional statement in Python?

Conditional statements are used to make decisions based on conditions.
The main conditional statements are `if`, `elif`, and `else`.

### Q2. What is the difference between a for loop and a while loop?

A `for` loop is generally used to iterate over a sequence or range, while a `while` loop repeats a block of code as long as a condition is True.

### Q3. What is the difference between break and continue?

`break` stops the loop completely, while `continue` skips the current iteration and moves to the next iteration.

### Q4. What is range()?

`range()` generates a sequence of numbers and is commonly used with a `for` loop to control iteration.

### Q5. What is the difference between if, elif, and else?

`if` checks the first condition, `elif` checks another condition if needed, and `else` executes when none of the previous conditions are True.
