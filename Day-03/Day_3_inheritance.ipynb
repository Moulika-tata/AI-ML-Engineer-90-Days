# Day 3 — OOP: Inheritance

## Topic 1: Introduction to Inheritance

### What is Inheritance?
Inheritance is an Object-Oriented Programming (OOP) concept that allows a child class to reuse attributes and methods from a parent class.

### Important Terms
- **Parent class:** The class whose features are inherited.
- **Child class:** The class that inherits features from the parent class.
- **Code reusability:** Reusing existing code instead of writing the same code again.

### Benefits of Inheritance
1. Reduces code duplication.
2. Improves code reusability.
3. Organizes related classes.
4. Allows child classes to add their own features.

### Syntax
`class Child(Parent):`

```python

# Topic 1: Introduction to Inheritance

class Animal:
    def eat(self):
        print("This animal eats food.")

class Dog(Animal):
    def bark(self):
        print("The dog barks.")

dog1 = Dog()
dog1.eat()
dog1.bark()
```

## Topic 2: Single Inheritance

Single inheritance occurs when one child class inherits from one parent class.

### Example
- Parent class: `Person`
- Child class: `Student`

The `Student` class inherits the `name` attribute and `introduce()` method from `Person`. It can also define its own attributes and methods.

### Syntax
`class Student(Person):`

### Benefit
Single inheritance allows a child class to reuse code from one parent class.

```python

# Topic 2: Single Inheritance

class Person:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        print("My name is", self.name)

class Student(Person):
    def __init__(self, name, student_id):
        super().__init__(name)
        self.student_id = student_id

    def display_id(self):
        print("Student ID:", self.student_id)

student1 = Student("Moulika", 101)
student1.introduce()
student1.display_id()
```

## Topic 3: The `super()` Function

The `super()` function allows a class to access the next applicable implementation in Python's Method Resolution Order (MRO).

### Uses
- Call the parent constructor using `super().__init__()`.
- Reuse a parent's method when overriding it.
- Avoid duplicating initialization code.

### Important Point
When a child class defines its own `__init__()` method, the parent's constructor is not automatically called. Use `super().__init__()` when parent initialization is needed.

```python

# Topic 3: Using super()

class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
    def __init__(self, name, student_id):
        super().__init__(name)
        self.student_id = student_id

student1 = Student("Moulika", 101)

print("Name:", student1.name)
print("Student ID:", student1.student_id)
```

## Topic 4: Method Overriding

Method overriding occurs when a child class defines its own implementation of a method already defined in the parent class.

### Why do we use it?
1. To customize inherited behavior.
2. To allow child classes to behave differently.
3. To make programs more flexible.

### Example
The `Animal` class defines a `sound()` method. The `Dog` and `Cat` classes override it to provide their own sounds.

```python

# Topic 4: Method Overriding

class Animal:
    def sound(self):
        print("Animals make different sounds.")

class Dog(Animal):
    def sound(self):
        print("Dog barks.")

class Cat(Animal):
    def sound(self):
        print("Cat meows.")

animal1 = Animal()
dog1 = Dog()
cat1 = Cat()

animal1.sound()
dog1.sound()
cat1.sound()
```

## Topic 5: Multilevel Inheritance

Multilevel inheritance occurs when inheritance happens across multiple levels.

### Example
`Animal → Dog → Puppy`

- `Dog` inherits from `Animal`.
- `Puppy` inherits from `Dog`.
- A `Puppy` object can use methods inherited through both levels.

### Benefit
Multilevel inheritance allows classes to extend functionality gradually.

```python

# Topic 5: Multilevel Inheritance

class Animal:
    def eat(self):
        print("Animal eats food.")

class Dog(Animal):
    def bark(self):
        print("Dog barks.")

class Puppy(Dog):
    def weep(self):
        print("Puppy makes a small cry.")

puppy1 = Puppy()

puppy1.eat()
puppy1.bark()
puppy1.weep()
```

## Topic 6: Hierarchical Inheritance

Hierarchical inheritance occurs when two or more child classes inherit from the same parent class.

### Example
- `Animal` is the parent class.
- `Dog` and `Cat` are child classes.
- Both child classes inherit the `eat()` method from `Animal`.

### Benefit
Common functionality can be written once in the parent class and reused by multiple child classes.

```python

# Topic 6: Hierarchical Inheritance

class Animal:
    def eat(self):
        print("This animal eats food.")

class Dog(Animal):
    def bark(self):
        print("Dog barks.")

class Cat(Animal):
    def meow(self):
        print("Cat meows.")

dog1 = Dog()
cat1 = Cat()

dog1.eat()
dog1.bark()

cat1.eat()
cat1.meow()
```

## Topic 7: Multiple Inheritance

Multiple inheritance occurs when one child class inherits from two or more parent classes.

### Example
- `Camera` provides a `take_photo()` method.
- `Phone` provides a `make_call()` method.
- `SmartPhone` inherits from both classes.

### Syntax
`class SmartPhone(Camera, Phone):`

### Benefit
The child class can reuse features from multiple parent classes.

```python

# Topic 7: Multiple Inheritance

class Camera:
    def take_photo(self):
        print("Taking a photo.")

class Phone:
    def make_call(self):
        print("Making a phone call.")

class SmartPhone(Camera, Phone):
    def use_internet(self):
        print("Browsing the internet.")

mobile = SmartPhone()

mobile.take_photo()
mobile.make_call()
mobile.use_internet()
```

## Topic 8: Hybrid Inheritance

Hybrid inheritance combines two or more inheritance types in one class hierarchy.

### Example
- `Student` and `Athlete` both inherit from `Person`.
- `StudentAthlete` inherits from both `Student` and `Athlete`.

This combines hierarchical inheritance with multiple inheritance.

Python uses Method Resolution Order (MRO) to determine how methods are searched in an inheritance hierarchy.

```python

# Topic 8: Hybrid Inheritance

class Person:
    def introduce(self):
        print("I am a person.")

class Student(Person):
    def study(self):
        print("I am studying.")

class Athlete(Person):
    def play(self):
        print("I am playing a sport.")

class StudentAthlete(Student, Athlete):
    pass

person1 = StudentAthlete()

person1.introduce()
person1.study()
person1.play()

print([cls.__name__ for cls in StudentAthlete.mro()])
```

## Topic 9: Method Resolution Order (MRO)

Method Resolution Order (MRO) is the order in which Python searches classes for a method or attribute.

### Important Points
- Python checks the child class first.
- If the method is not found, Python searches the next class in the MRO.
- `ClassName.mro()` displays the resolution order.
- MRO is particularly important in multiple and hybrid inheritance.

```python

# Topic 9: Method Resolution Order

class Animal:
    def sound(self):
        print("Animal sound.")

class Dog(Animal):
    def sound(self):
        print("Dog barks.")

dog1 = Dog()
dog1.sound()

print([cls.__name__ for cls in Dog.mro()])
```

## Topic 10: Using super() with Method Overriding

A child class can override a parent method and still reuse the parent's behavior by calling `super()`.

### Example
The `Dog` class calls `Animal.sound()` first and then prints its own sound.

### Benefit
This avoids repeating the parent's code and allows the child to extend existing behavior.

```python

# Topic 10: super() with Method Overriding

class Animal:
    def sound(self):
        print("Animal makes a sound.")

class Dog(Animal):
    def sound(self):
        super().sound()
        print("Dog barks.")

dog1 = Dog()
dog1.sound()
```
