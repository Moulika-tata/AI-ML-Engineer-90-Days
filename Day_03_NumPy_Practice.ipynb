## Vectorized Operations

Vectorization allows NumPy to perform operations on an entire array without using an explicit loop.

Examples:
- Addition
- Subtraction
- Multiplication
- Division

```python

arr = np.array([10, 20, 30, 40, 50])

print("Addition:", arr + 5)
print("Subtraction:", arr - 5)
print("Multiplication:", arr * 2)
print("Division:", arr / 10)
```

## Aggregation Functions

Aggregation functions summarize numerical data.

- `np.sum()` — calculates the total.
- `np.mean()` — calculates the average.
- `np.min()` — finds the minimum value.
- `np.max()` — finds the maximum value.
- `np.std()` — calculates standard deviation.
- `np.var()` — calculates variance.

```python

arr = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Minimum:", np.min(arr))
print("Maximum:", np.max(arr))
print("Standard deviation:", np.std(arr))
print("Variance:", np.var(arr))
```

## Array Slicing

Slicing extracts a portion of an array.

Syntax: `array[start:stop]`

- The start index is included.
- The stop index is excluded.
- Python indexing begins at zero.

```python

arr = np.array([5, 10, 15, 20, 25, 30])

print("First three elements:", arr[:3])
print("Elements at indices 1 to 3:", arr[1:4])
print("Last three elements:", arr[-3:])
```

## Boolean Filtering

Boolean filtering selects elements that satisfy a condition.

- `arr > 30` selects values greater than 30.
- `arr < 30` selects values less than 30.
- `&` combines multiple conditions using AND.
- Each condition must be enclosed in parentheses when using `&`.

```python

arr = np.array([12, 25, 38, 45, 60, 67, 75])

print("Values greater than 40:", arr[arr > 40])
print("Values less than 50:", arr[arr < 50])
print("Values between 40 and 70:", arr[(arr > 40) & (arr < 70)])
```

## NumPy Broadcasting

Broadcasting allows NumPy to perform operations between arrays with compatible shapes.

Examples:
- Adding a scalar to every element.
- Adding a one-dimensional array to each row of a compatible two-dimensional array.

```python

matrix = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("Matrix + 5:")
print(matrix + 5)

values = np.array([10, 20, 30])

print("Matrix + values:")
print(matrix + values)
```

## Understanding Axis

For a two-dimensional NumPy array:

- `axis=0` performs operations down the rows, producing one result per column.
- `axis=1` performs operations across columns, producing one result per row.

Examples:
- `np.sum(arr, axis=0)` calculates column totals.
- `np.mean(arr, axis=1)` calculates row averages.

```python

marks = np.array([
    [70, 80, 90],
    [60, 75, 85],
    [88, 92, 78]
])

print("Column totals (axis=0):")
print(np.sum(marks, axis=0))

print("Column averages (axis=0):")
print(np.mean(marks, axis=0))

print("Row totals (axis=1):")
print(np.sum(marks, axis=1))

print("Row averages (axis=1):")
print(np.mean(marks, axis=1))
```

## Random Data Generation

NumPy provides functions to generate random data.

- `np.random.rand()` generates random floating-point numbers between 0 and 1.
- `np.random.randint()` generates random integers.
- The upper bound of `np.random.randint()` is excluded.
- `size` specifies the number or shape of generated values.

```python

# Generate 5 random floating-point numbers
print("Random floats:")
print(np.random.rand(5))

# Generate 10 random integers from 1 to 100
random_numbers = np.random.randint(1, 101, size=10)
print("Random integers:")
print(random_numbers)

# Generate a 3x4 random matrix
random_matrix = np.random.randint(10, 51, size=(3, 4))
print("Random matrix:")
print(random_matrix)

print("Shape:", random_matrix.shape)
print("Mean:", np.mean(random_matrix))
```

## Array Reshaping

The `reshape()` method changes an array's dimensions without changing its elements.

The new shape must contain the same total number of elements as the original array.

```python

arr = np.arange(1, 7)

print("Original array:")
print(arr)

reshaped = arr.reshape(2, 3)

print("Reshaped array:")
print(reshaped)

print("Shape:", reshaped.shape)
```
