## 1. HCF — Highest Common Factor

HCF is the largest number that divides two or more numbers exactly.

### Example

Find the HCF of 12 and 18.

Factors of 12:
1, 2, 3, 4, 6, 12

Factors of 18:
1, 2, 3, 6, 9, 18

Common factors:
1, 2, 3, 6

Therefore:

HCF = 6

## 2. LCM — Least Common Multiple

LCM is the smallest positive number that is a multiple of two or more numbers.

### Example

Find the LCM of 4 and 6.

Multiples of 4:
4, 8, 12, 16, 20, ...

Multiples of 6:
6, 12, 18, 24, ...

First common multiple:

LCM = 12

## 3. HCF vs LCM

### HCF
- Uses factors
- Finds the highest common factor
- HCF → Common prime factors + Lowest powers

### LCM
- Uses multiples
- Finds the least common multiple
- LCM → All prime factors + Highest powers

### Easy Memory Trick

HCF → Highest Factor

LCM → Least Multiple

## 4. Prime Factorization

Prime factorization means expressing a number as a product of prime numbers.

### Example 1

12 = 2 × 2 × 3

12 = 2² × 3

### Example 2

18 = 2 × 3 × 3

18 = 2 × 3²

### Example 3

30 = 2 × 3 × 5

## 5. HCF Using Prime Factorization

Example:

Find HCF of 12 and 18.

12 = 2² × 3

18 = 2 × 3²

Common prime factors:
2 and 3

Take the lowest powers:

2¹ × 3¹

HCF = 2 × 3

HCF = 6

## 6. LCM Using Prime Factorization

Example:

Find LCM of 12 and 18.

12 = 2² × 3

18 = 2 × 3²

Take all prime factors with their highest powers:

2² × 3²

LCM = 4 × 9

LCM = 36

## 7. Important HCF & LCM Formula

For two positive integers:

HCF × LCM = First Number × Second Number

### Example

Numbers = 12 and 18

HCF = 6
LCM = 36

6 × 36 = 216

12 × 18 = 216

Therefore:

HCF × LCM = Product of the two numbers

```python
def prime_factors(n):
    factors = []
    divisor = 2

    while n > 1:
        while n % divisor == 0:
            factors.append(divisor)
            n //= divisor
        divisor += 1

    return factors


number = int(input("Enter a number: "))

print("Prime factors:", prime_factors(number))
```

```python
import math

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

hcf = math.gcd(a, b)

print("HCF =", hcf)
```

```python
import math

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

lcm = math.lcm(a, b)

print("LCM =", lcm)
```

## 8. HCF & LCM Word Problems

### LCM Pattern

Use LCM when events repeat and we need to find when they happen together again.

Examples:
- Bells ringing
- Buses arriving
- Machines repeating cycles
- Lights blinking

### HCF Pattern

Use HCF when we need the maximum number of identical groups or sets.

Examples:
- Maximum identical packets
- Maximum identical groups
- Dividing items equally
