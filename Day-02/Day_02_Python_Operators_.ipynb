## What is an Operator?

An operator is a symbol or keyword used to perform an operation on values or variables.

Example:

a = 10
b = 5

a + b

Here:
- a and b are operands
- + is the operator
- a + b is an expression

## 1. Arithmetic Operators

Arithmetic operators are used to perform mathematical calculations.

+   → Addition
-   → Subtraction
*   → Multiplication
/   → Division
//  → Floor Division
%   → Modulo / Remainder
**  → Exponentiation

Examples:

10 + 5 = 15
10 - 5 = 5
10 * 5 = 50
10 / 5 = 2.0
10 // 3 = 3
10 % 3 = 1
2 ** 3 = 8

```python
a = 10
b = 3

print("Addition =", a + b)
print("Subtraction =", a - b)
print("Multiplication =", a * b)
print("Division =", a / b)
print("Floor Division =", a // b)
print("Remainder =", a % b)
print("Power =", a ** b)
```

## 2. Comparison Operators

Comparison operators are used to compare two values.

The result is always True or False.

>   → Greater than
<   → Less than
>=  → Greater than or equal to
<=  → Less than or equal to
==  → Equal to
!=  → Not equal to

Important:

=  → Assignment
== → Comparison

```python
a = 10
b = 5

print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
print(a == b)
print(a != b)
```

## 3. Logical Operators

Logical operators are used to combine or reverse conditions.

and → Both conditions must be True
or  → At least one condition must be True
not → Reverses the Boolean value

Examples:

True and True → True
True and False → False

True or False → True
False or False → False

not True → False
not False → True

```python
print(True and True)
print(True and False)

print(True or False)
print(False or False)

print(not True)
print(not False)
```

## 4. Assignment Operators

Assignment operators are used to assign or update values in variables.

=    → Assignment
+=   → Add and assign
-=   → Subtract and assign
*=   → Multiply and assign
/=   → Divide and assign
//=  → Floor divide and assign
%=   → Modulo and assign
**=  → Power and assign

Examples:

x += 5
means
x = x + 5

x *= 2
means
x = x * 2

```python
x = 10

x += 5
print("After += :", x)

x -= 3
print("After -= :", x)

x *= 2
print("After *= :", x)

x //= 2
print("After //= :", x)

x %= 3
print("After %= :", x)
```

## 5. Operator Precedence

Python follows an order when evaluating expressions.

Order:

1. Parentheses ()
2. Exponentiation **
3. Multiplication *, Division /, Floor Division //, Modulo %
4. Addition +, Subtraction -

Example:

2 + 3 * 4

First:
3 * 4 = 12

Then:
2 + 12 = 14

Therefore:
2 + 3 * 4 = 14

Parentheses can change the order:

(2 + 3) * 4 = 20

```python
print(2 + 3 * 4)
print((2 + 3) * 4)
print(2 ** 3 + 1)
print(10 - 2 * 3)
```

## Program 1 — Calculate Total and Average

Take three numbers from the user.

Calculate:
- Total
- Average

```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
num3 = int(input("Enter third number: "))

total = num1 + num2 + num3
average = total / 3

print("Total =", total)
print("Average =", average)
```

## Program 2 — Product and Remainder

Take two integers from the user.

Calculate:
- Product
- Remainder

```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

product = num1 * num2
remainder = num1 % num2

print("Product =", product)
print("Remainder =", remainder)
```

## Program 3 — Divisibility by 5

Check whether a number is divisible by 5.

```python
num = int(input("Enter number: "))

if num % 5 == 0:
    print("Divisible by 5")
else:
    print("Not divisible by 5")
```

## Program 4 — Positive, Negative, or Zero

Check whether a number is:
- Positive
- Negative
- Zero

```python
num = int(input("Enter number: "))

if num > 0:
    print("Positive")
elif num < 0:
    print("Negative")
else:
    print("Zero")
```

## Program 5 — Compare Two Numbers

Take two numbers and determine:
- First number is greater
- Second number is greater
- Both numbers are equal

```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

if num1 > num2:
    print("First number is greater")
elif num1 == num2:
    print("Both numbers are equal")
else:
    print("Second number is greater")
```

## Program 6 — Even and Divisible by 5

Check whether a number is both:
- Even
- Divisible by 5

Both conditions must be True.

```python
num = int(input("Enter number: "))

if num % 2 == 0 and num % 5 == 0:
    print("Even and divisible by 5")
else:
    print("Condition not satisfied")
```

## Program 7 — Eligibility Check

A person is eligible if:

Age >= 18
AND
Marks >= 60

```python
age = int(input("Enter age: "))
marks = int(input("Enter marks: "))

if age >= 18 and marks >= 60:
    print("Eligible")
else:
    print("Not Eligible")
```

## Program 8 — Discount Calculator

If the price is 1000 or more:
- Give a 10% discount.

Otherwise:
- No discount.

```python
price = int(input("Enter price: "))

if price >= 1000:
    discount = price * 10 / 100
    final_price = price - discount

    print("Discount =", discount)
    print("Final price =", final_price)
else:
    print("Discount = 0")
    print("Final price =", price)
```

## Interview Questions

### 1. What is an operator in Python?

An operator is a symbol or keyword used to perform an operation on values or variables.

### 2. What is the difference between /, //, and %?

/ performs normal division and returns a floating-point result.

// performs floor division and returns the floor value.

% returns the remainder of a division.

Example:

10 / 3  → 3.333...
10 // 3 → 3
10 % 3  → 1

### 3. What is the difference between = and ==?

= is used for assignment.

== is used to compare two values for equality.
