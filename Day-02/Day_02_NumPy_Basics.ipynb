## 1. What is NumPy?

NumPy is a Python library used for numerical computing.

It provides powerful array structures and efficient mathematical operations.

NumPy is widely used in:
- Artificial Intelligence
- Machine Learning
- Data Science
- Deep Learning

NumPy arrays are generally more suitable than normal Python lists for
numerical and mathematical operations.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr)
```

## 2. NumPy Array Properties

Important NumPy array properties:

- shape → tells us the number of rows and columns
- ndim → tells us the number of dimensions
- size → tells us the total number of elements
- dtype → tells us the data type of the elements

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Array:")
print(arr)

print("Shape:", arr.shape)
print("Dimensions:", arr.ndim)
print("Size:", arr.size)
print("Data type:", arr.dtype)
```

## 3. Indexing

Indexing is used to access individual elements of an array.

Python and NumPy use zero-based indexing.

Therefore:
- First element → index 0
- Second element → index 1
- Third element → index 2

For a 2D array, we use:

array[row, column]

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Element at row 0, column 0:", arr[0, 0])
print("Element at row 0, column 1:", arr[0, 1])
print("Element at row 1, column 0:", arr[1, 0])
print("Element at row 1, column 2:", arr[1, 2])
```

```python
print("Last element:", arr[-1, -1])
print("Second last column of last row:", arr[-1, -2])
```

## 4. Slicing

Slicing is used to access multiple elements from an array.

Basic syntax:

array[start:stop]

The start index is included and the stop index is excluded.

For a 2D array:

array[row_start:row_stop, column_start:column_stop]

```python
arr = np.array([10, 20, 30, 40, 50])

print("Elements from index 1 to 3:", arr[1:4])
print("First three elements:", arr[:3])
print("Elements from index 2:", arr[2:])
print("Reverse array:", arr[::-1])
```

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("First two rows:")
print(arr[:2])

print("Second column:")
print(arr[:, 1])

print("First two columns:")
print(arr[:, :2])

print("Rows 1 onward and columns 1 onward:")
print(arr[1:, 1:])
```

## 5. Vectorized Operations

Vectorization allows NumPy to perform operations on an entire array
without manually writing a loop for every element.

Examples:
- Addition
- Subtraction
- Multiplication
- Division
- Power
- Comparisons

```python
arr = np.array([10, 20, 30, 40, 50])

print("Original:", arr)
print("Add 5:", arr + 5)
print("Subtract 5:", arr - 5)
print("Multiply by 2:", arr * 2)
print("Divide by 10:", arr / 10)
print("Square:", arr ** 2)
print("Greater than 25:", arr > 25)
```

## 6. Broadcasting

Broadcasting is a NumPy feature that allows arithmetic operations between
arrays of compatible shapes without explicitly writing loops.

For example, when a scalar is used with an array, NumPy applies the
operation to every element of the array.

```python
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Add 5:")
print(arr + 5)

print("Multiply by 2:")
print(arr * 2)

print("Subtract 10:")
print(arr - 10)
```

```python
values = np.array([1, 2, 3])

print(arr + values)
```

## 7. Reshape

reshape() changes the structure or shape of an array without changing
the actual values.

The total number of elements must remain the same.

For example:

6 elements can be reshaped into:
- (2, 3)
- (3, 2)
- (6, 1)

because each shape contains 6 elements.

```python
arr = np.array([10, 20, 30, 40, 50, 60])

print("Original array:")
print(arr)

reshaped = arr.reshape(2, 3)

print("Reshaped array:")
print(reshaped)

print("New shape:", reshaped.shape)
```

```python
arr = np.array([10, 20, 30, 40, 50, 60])

reshaped = arr.reshape(3, 2)

print(reshaped)
print("Shape:", reshaped.shape)
```

### Reshape Rule

The number of elements before and after reshaping must be the same.

Example:

2 × 3 = 6 → valid

3 × 2 = 6 → valid

4 × 2 = 8 → invalid for an array containing only 6 elements.

## 8. NumPy Statistical Functions

Important functions:

- np.mean() → calculates the average
- np.max() → finds the largest value
- np.min() → finds the smallest value
- np.sum() → calculates the total

```python
arr = np.array([10, 20, 30, 40, 50])

print("Mean:", np.mean(arr))
print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))
print("Sum:", np.sum(arr))
```

## 9. Filtering

Filtering is used to select elements that satisfy a condition.

Basic syntax:

array[condition]

For example:

arr[arr > 30]

This checks every element and returns the actual values for which
the condition is True.

```python
arr = np.array([10, 20, 30, 40, 50, 60])

print("Greater than 30:")
print(arr[arr > 30])

print("Less than 40:")
print(arr[arr < 40])

print("Equal to 50:")
print(arr[arr == 50])

print("Greater than or equal to 30:")
print(arr[arr >= 30])
```

## 10. Filtering with Multiple Conditions

NumPy allows multiple conditions.

Symbols:

& → AND
| → OR

Each condition should be placed inside parentheses.

```python
arr = np.array([10, 20, 30, 40, 50, 60])

print("Greater than 20 AND less than 50:")
print(arr[(arr > 20) & (arr < 50)])

print("Less than 20 OR greater than 50:")
print(arr[(arr < 20) | (arr > 50)])

print("Greater than or equal to 30 AND less than or equal to 50:")
print(arr[(arr >= 30) & (arr <= 50)])
```

## 11. Matrix Multiplication

Started learning matrix multiplication.

Key difference:

* → element-wise multiplication
@ → matrix multiplication

Matrix multiplication will be completed in the next session.

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Element-wise multiplication:")
print(A * B)

print("Matrix multiplication:")
print(A @ B)
```

## 12. NumPy Arrays vs Python Lists

Python lists are general-purpose collections.

NumPy arrays are designed for numerical computing.

NumPy supports vectorized operations, multidimensional arrays,
and mathematical operations directly.

For AI/ML and numerical datasets, NumPy is generally more efficient
and convenient than normal Python lists.

```python
numbers = [10, 20, 30, 40, 50]

result = []

for number in numbers:
    result.append(number + 5)

print("Python list result:", result)
```

```python
import numpy as np
numbers = np.array([10, 20, 30, 40, 50])

print("NumPy result:", numbers + 5)
```
