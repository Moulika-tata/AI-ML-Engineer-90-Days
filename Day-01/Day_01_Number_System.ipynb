## 1. Number Types

### Natural Numbers
1, 2, 3, 4, ...

### Whole Numbers
0, 1, 2, 3, 4, ...

### Integers
..., -3, -2, -1, 0, 1, 2, 3, ...

### Key Difference
Natural → starts from 1
Whole → starts from 0
Integers → include negative numbers, zero and positive numbers

## 2. Factors and Multiples

### Factors
A factor divides a number exactly with remainder 0.

Example:
Factors of 12 = 1, 2, 3, 4, 6, 12

### Multiples
Multiples are obtained by multiplying a number by 1, 2, 3, ...

Example:
Multiples of 5 = 5, 10, 15, 20, 25, ...

## 3. Prime and Composite Numbers

### Prime Number
A number having exactly two factors:
1 and itself.

Examples:
2, 3, 5, 7, 11, 13

### Composite Number
A number having more than two factors.

Examples:
4, 6, 8, 9, 10, 12

### Important
1 is neither prime nor composite.

2 is the only even prime number.

## 4. Divisibility Rules

### Divisible by 2
Last digit must be 0, 2, 4, 6 or 8.

### Divisible by 3
Sum of digits must be divisible by 3.

### Divisible by 4
Last two digits must be divisible by 4.

### Divisible by 5
Last digit must be 0 or 5.

### Divisible by 6
Number must be divisible by both 2 and 3.

### Divisible by 9
Sum of digits must be divisible by 9.

### Divisible by 10
Last digit must be 0.
