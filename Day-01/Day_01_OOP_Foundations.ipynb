## 1. What is OOP?

OOP stands for Object-Oriented Programming.

It is a programming approach where programs are organized around objects.

An object contains:

- Data → Attributes
- Behavior → Methods

### Simple Example

A Student object can have:

- Name
- Age
- CGPA

And can perform actions such as:

- Study
- Attend class
- Write an exam

## 2. What is a Class?

A class is a blueprint or template used to create objects.

### Example

A `Student` class can define:

- Name
- Age
- CGPA
- Study behavior

The class itself is the blueprint. Objects are created from the class.

````python
class Student:
    pass
````

## 3. What is an Object?

An object is an instance of a class.

One class can be used to create multiple objects.

Example:

- `Student` → Class
- `student1` → Object
- `student2` → Object

````python
class Student:
    pass

student1 = Student()
student2 = Student()

print(student1)
print(student2)
````

## 4. What are Attributes?

Attributes are the data or properties of an object.

For a Student object, examples of attributes are:

- Name
- Age
- CGPA
- College

````python
class Student:
    def __init__(self, name, age, cgpa):
        self.name = name
        self.age = age
        self.cgpa = cgpa

student1 = Student("mouli", 21, 8.4)

print(student1.name)
print(student1.age)
print(student1.cgpa)
````

## 5. What is a Method?

A method is a function defined inside a class.

Methods represent the behavior or actions of an object.

Example:

A Student can:

- Study
- Attend class
- Write an exam

````python
class Student:
    def study(self):
        print("Student is studying")

student1 = Student()
student1.study()
````

## 8. Class vs Object

| Class | Object |
|---|---|
| Blueprint/template | Instance of a class |
| Defines structure and behavior | Represents a specific instance |
| Example: `Student` | Example: `student1` |

# Practice 1 — Car Class

Create a `Car` class with:

- brand
- model
- price

Method:

- `drive()`

````python
class Car:
    def __init__(self, brand, model, price):
        self.brand = brand
        self.model = model
        self.price = price

    def drive(self):
        print(self.brand, self.model, "is driving")


car1 = Car("Toyota", "Camry", 30000)
car1.drive()
````

# Practice 2 — Book Class

Create a `Book` class with:

- title
- author
- price

Method:

- `display()`

````python
class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price

    def display(self):
        print("Title:", self.title)
        print("Author:", self.author)
        print("Price:", self.price)


book1 = Book("Python Programming", "XYZ", 500)
book1.display()
````

# Practice 3 — BankAccount Class

Create a `BankAccount` class with:

### Attributes
- account_holder
- balance

### Methods
- deposit()
- withdraw()
- display_balance()

````python
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount

    def withdraw(self, amount):
        self.balance = self.balance - amount

    def display_balance(self):
        print(self.balance)


account = BankAccount("Mouli", 5000)

account.deposit(1000)
account.withdraw(500)

account.display_balance()
````
